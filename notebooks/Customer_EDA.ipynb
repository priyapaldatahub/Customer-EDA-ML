```python
#12. Separate Categorical and Numerical Columns
```

```python
categorical_columns = ['Channel', 'Region']

numerical_columns = [
    'Fresh',
    'Milk',
    'Grocery',
    'Frozen',
    'Detergents_Paper',
    'Delicatessen'
]
```

```python
#13. Analyze Categorical Variables
```

```python
#Channel Distribution
sns.countplot(x='Channel', data=df)
plt.title('Channel Distribution')
plt.show()
```

```python
#Region Distribution
sns.countplot(x='Region', data=df)
plt.title('Region Distribution')
plt.show()
```

```python
df['Channel'].value_counts()
```

```python
df['Region'].value_counts()
```

```python
#14. Distribution of Numerical Variables
df[numerical_columns].hist(figsize=(12, 8))
plt.tight_layout()
plt.show()
```

```python
#15. Detect Skewness

df[numerical_columns].skew()
```

```python
for col in numerical_columns:
    sns.histplot(df[col], kde=True)
    plt.title(f'Distribution of {col}')
    plt.show()
```

```python
#16. Detect Outliers Using Boxplots
```

```python
for col in numerical_columns:
    sns.boxplot(x=df[col])
    plt.title(f'Boxplot of {col}')
    plt.show()
```

```python
#17. Correlation Analysis
plt.figure(figsize=(10, 7))


sns.heatmap(
    df.corr(),
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)


plt.title('Correlation Heatmap')
plt.show()
```

```python
#18. Check Pairwise Relationships
sns.pairplot(df[numerical_columns])
plt.show()
```

```python
#19. Check Unique Values
for col in df.columns:
    print(col, ":", df[col].nunique())
```
